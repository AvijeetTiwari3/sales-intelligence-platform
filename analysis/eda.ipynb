```python
import sqlite3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="darkgrid")

# Connect to Analytics Data Warehouse
conn = sqlite3.connect("../database/sales.db")
df = pd.read_sql("SELECT * FROM orders;", conn)
conn.close()

print(f"Loaded {len(df):,} transactions with {df.shape[1]} engineered features.")
df.head(3)
```

## 1. Top-Line Financial & Operational Metrics
- Total Gross Merchandise Value (GMV)
- Net Operating Margin
- Transactional Density

```python
total_gmv = df["sales"].sum()
net_profit = df["profit"].sum()
blended_margin = (net_profit / total_gmv) * 100

print(f"Gross Revenue:   ${total_gmv:,.2f}")
print(f"Net Profit:      ${net_profit:,.2f}")
print(f"Blended Margin:  {blended_margin:.2f}%")
print(f"Unique Customers: {df['customer_id'].nunique()}")
print(f"Total Orders:     {df['order_id'].nunique()}")
```

## 2. Pricing Fallacy: The Profit Margin Destruction Curve
How do discount tiers impact net profit generation? Let's quantify revenue leakage.

```python
discount_summary = df.groupby("discount_tier").agg(
    total_sales=("sales", "sum"),
    total_profit=("profit", "sum"),
    order_volume=("order_id", "count"),
    average_margin=("profit_margin", lambda x: np.round(x.mean() * 100, 2))
).reset_index()

discount_summary
```

## 3. Customer Concentration Risk (Pareto Principle)
Does 20% of the customer base drive 80% of revenue?

```python
cust_revenue = df.groupby("customer_id")["sales"].sum().sort_values(ascending=False)
top_20_count = int(len(cust_revenue) * 0.20)
top_20_share = (cust_revenue.iloc[:top_20_count].sum() / total_gmv) * 100

print(f"Top 20% of accounts ({top_20_count} customers) contribute {top_20_share:.2f}% of total sales.")
```
